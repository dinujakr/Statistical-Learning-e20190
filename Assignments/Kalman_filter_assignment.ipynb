# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

### 1. Distribution of the Predicted State

Since the predicted state $x^-_k$ is a linear combination of independent Gaussian random variables ($x^+_{k-1}$ and $w_{k-1}$), it is also Gaussian. To completely specify its distribution $\mathcal{N}(m^-_k, P^-_k)$, we only need to derive its **mean** and **covariance**.

#### Finding the Predicted Mean ($m^-_k$)

Using the linearity of the expectation operator $E[\cdot]$:

$$m^-_k = E[x^-_k]$$
$$m^-_k = E[A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1}]$$
$$m^-_k = A_{k-1}E[x^+_{k-1}] + G_{k-1}E[w_{k-1}]$$

Given that $E[x^+_{k-1}] = m_{k-1}$ and the process noise is zero-mean ($E[w_{k-1}] = 0$):

$$m^-_k = A_{k-1}m_{k-1}$$

---

#### Finding the Predicted Covariance ($P^-_k$)

The covariance of $x^-_k$ is defined by the variance operator $\text{Var}(\cdot)$:

$$P^-_k = \text{Var}(x^-_k)$$
$$P^-_k = \text{Var}(A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1})$$

Because the prior state $x^+_{k-1}$ and the process noise $w_{k-1}$ are independent, their cross-covariance terms vanish. Therefore, the variance of their sum is simply the sum of their individual variances:

$$P^-_k = \text{Var}(A_{k-1}x^+_{k-1}) + \text{Var}(G_{k-1}w_{k-1})$$

Applying the matrix scaling property $\text{Var}(BX) = B\text{Var}(X)B^T$ for a constant matrix $B$:

$$P^-_k = A_{k-1}\text{Var}(x^+_{k-1})A_{k-1}^T + G_{k-1}\text{Var}(w_{k-1})G_{k-1}^T$$

Substituting the known variances $\text{Var}(x^+_{k-1}) = P_{k-1}$ and $\text{Var}(w_{k-1}) = \Sigma_p$:

$$P^-_k = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T$$

---

#### Conclusion

Combining both derivations, the distribution of the predicted state is given by:

$$x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$$

### 2. Distribution of the Predicted Measurement

We are given the measurement model equation:
$$y^-_k = H_k x^-_k + z_k$$

where:
* The predicted state is $x^-_k \sim \mathscr{N}(m^-_k, P^-_k)$
* The measurement noise is $z_k \sim \mathscr{N}(0, \Sigma_m)$
* $x^-_k$ and $z_k$ are assumed to be mutually independent.

Since $y^-_k$ is a linear combination of independent Gaussian random variables, it is also Gaussian. To completely specify its distribution, we derive its **mean** and **covariance**.

#### Finding the Predicted Measurement Mean

Using the linearity property of the expectation operator $\mathbb{E}[\cdot]$:

$$\mathbb{E}[y^-_k] = \mathbb{E}[H_k x^-_k + z_k]$$
$$\mathbb{E}[y^-_k] = H_k \mathbb{E}[x^-_k] + \mathbb{E}[z_k]$$

Substituting the known expected values $\mathbb{E}[x^-_k] = m^-_k$ and $\mathbb{E}[z_k] = 0$:

$$\mathbb{E}[y^-_k] = H_k m^-_k$$

---

#### Finding the Predicted Measurement Covariance

The covariance matrix of $y^-_k$ is given by the variance operator $\text{Var}(\cdot)$:

$$\text{Var}(y^-_k) = \text{Var}(H_k x^-_k + z_k)$$

Because the predicted state $x^-_k$ and the measurement noise $z_k$ are independent, their cross-covariance terms vanish. Therefore, the variance of their sum equals the sum of their individual variances:

$$\text{Var}(y^-_k) = \text{Var}(H_k x^-_k) + \text{Var}(z_k)$$

Applying the linear transformation property of covariance matrices, $\text{Var}(BX) = B \text{Var}(X) B^T$ for a constant matrix $B$:

$$\text{Var}(y^-_k) = H_k \text{Var}(x^-_k) H_k^T + \text{Var}(z_k)$$

Substituting the known covariance matrices $\text{Var}(x^-_k) = P^-_k$ and $\text{Var}(z_k) = \Sigma_m$:

$$\text{Var}(y^-_k) = H_k P^-_k H_k^T + \Sigma_m$$

---

#### Conclusion

Combining the derived mean and covariance, the distribution of the predicted measurement is given by:

$$y^-_k \sim \mathscr{N}(H_k m^-_k, H_k P^-_k H_k^T + \Sigma_m)$$

### 3. Joint Distribution of the Predicted State and Measurement

We want to find the joint distribution of the stacked vector:

$$\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix}$$

We can express this stacked system as a linear transformation of the predicted state $x_k^-$ and the measurement noise $z_k$. From the measurement equation $y_k^- = H_k x_k^- + z_k$, we write the system in matrix form:

$$\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} = \begin{bmatrix} I \\ H_k \end{bmatrix} x_k^- + \begin{bmatrix} 0 \\ I \end{bmatrix} z_k$$

Since $x_k^-$ and $z_k$ are independent Gaussian random variables, their linear combination is also jointly Gaussian. We completely specify the distribution by finding the **joint mean** and the **joint covariance matrix**.

---

#### Finding the Joint Mean

Using the linearity of the expectation operator $\mathbb{E}[\cdot]$:

$$\mathbb{E}\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} = \begin{bmatrix} \mathbb{E}[x_k^-] \\ \mathbb{E}[y_k^-] \end{bmatrix}$$

From our previous derivations, we know that $\mathbb{E}[x_k^-] = m_k^-$ and $\mathbb{E}[y_k^-] = H_k m_k^-$. Therefore:

$$\mathbb{E}\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} = \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}$$

---

#### Finding the Joint Covariance Matrix

The joint covariance matrix of a stacked vector is structured as:

$$\text{Var}\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} = \begin{bmatrix} \text{Var}(x_k^-) & \text{Cov}(x_k^-, y_k^-) \\ \text{Cov}(y_k^-, x_k^-) & \text{Var}(y_k^-) \end{bmatrix}$$

We already know the diagonal blocks from the previous steps:
* $\text{Var}(x_k^-) = P_k^-$
* $\text{Var}(y_k^-) = H_k P_k^- H_k^T + \Sigma_m$

Now, we compute the cross-covariance block $\text{Cov}(x_k^-, y_k^-)$:

$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(y_k^- - H_k m_k^-)^T \right]$$

Substitute $y_k^- = H_k x_k^- + z_k$:

$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(H_k x_k^- + z_k - H_k m_k^-)^T \right]$$
$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-) \left( H_k(x_k^- - m_k^-) + z_k \right)^T \right]$$
$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(x_k^- - m_k^-)^T H_k^T \right] + \mathbb{E}\left[ (x_k^- - m_k^-) z_k^T \right]$$

Because the predicted state $x_k^-$ and the measurement noise $z_k$ are mutually independent, their cross-expectation is zero ($\mathbb{E}\left[ (x_k^- - m_k^-) z_k^T \right] = 0$). This leaves:

$$\text{Cov}(x_k^-, y_k^-) = \mathbb{E}\left[ (x_k^- - m_k^-)(x_k^- - m_k^-)^T \right] H_k^T$$
$$\text{Cov}(x_k^-, y_k^-) = P_k^- H_k^T$$

Since $\text{Cov}(y_k^-, x_k^-) = \left[\text{Cov}(x_k^-, y_k^-)\right]^T$, we get:

$$\text{Cov}(y_k^-, x_k^-) = (P_k^- H_k^T)^T = H_k P_k^-$$

---

#### Conclusion

Assembling the mean vector and the blocks of the covariance matrix yields the final joint distribution:

$$\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right)$$

### 4. Conditional Distribution of the Updated State (Measurement Update)

We want to find the conditional distribution of the state given an observed measurement:

$$x^{+}_k \triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$$

From Step 3, we established that the joint distribution of the predicted state and measurement is Gaussian:

$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right)$$

To simplify notation, let the joint covariance matrix blocks be denoted as:
* $\Sigma_{xx} = P_k^-$
* $\Sigma_{xy} = P_k^- H_k^T$
* $\Sigma_{yx} = H_k P_k^-$
* $\Sigma_{yy} = H_k P_k^- H_k^T + \Sigma_m$



#### Property of Conditional Gaussian Distributions

For any jointly Gaussian partitioned random vector, the conditional distribution $(X \mid Y = y)$ is also Gaussian:

$$(X \mid Y = y) \sim \mathscr{N}(\mu_{X \mid Y}, \Sigma_{X \mid Y})$$

where the conditional mean and conditional covariance are given by:
$$\mu_{X \mid Y} = \mu_x + \Sigma_{xy} \Sigma_{yy}^{-1} (y - \mu_y)$$
$$\Sigma_{X \mid Y} = \Sigma_{xx} - \Sigma_{xy} \Sigma_{yy}^{-1} \Sigma_{yx}$$


#### Deriving the Kalman Gain ($K_k$)

We define the Kalman Gain matrix $K_k$ as the coefficient term $\Sigma_{xy} \Sigma_{yy}^{-1}$:

$$K_k \triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$



#### Deriving the Updated Mean ($m_k$)

Applying the conditional mean formula with our partitioned blocks, where $y = y^{\mathrm{obs}}_{k}$:

$$m_k = \mu_{x \mid y} = m_k^- + \Sigma_{xy} \Sigma_{yy}^{-1} (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

Substituting the definition of $K_k$:

$$m_k = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$



#### Deriving the Updated Covariance ($P_k$)

Applying the conditional covariance formula:

$$P_k = \Sigma_{X \mid Y} = \Sigma_{xx} - \Sigma_{xy} \Sigma_{yy}^{-1} \Sigma_{yx}$$

Substitute $\Sigma_{xx} = P_k^-$, $\Sigma_{yx} = H_k P_k^-$, and use the definition of $K_k = \Sigma_{xy} \Sigma_{yy}^{-1}$:

$$P_k = P_k^- - K_k (H_k P_k^-)$$

Factoring out $P_k^-$ from the right-hand side yields:

$$P_k = (I - K_k H_k) P_k^-$$



#### Conclusion

Since the conditional distribution of a joint Gaussian distribution remains Gaussian, we have:

$$x^{+}_k \triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k)$$

where:
* $K_k \triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$
* $m_k = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$
* $P_k = (I - K_k H_k) P_k^-$

### 5. Evaluation of Conditional Expectation and Variance

From the derivation of the conditional distribution in Step 4, we defined the updated state vector as:

$$x^{+}_k \triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k)$$

By definition, the mean vector and covariance matrix of a Gaussian distribution correspond exactly to its first and second central moments (the expectation and variance). Therefore, we can directly read the values from the results in Step 4.


#### Finding the Conditional Expectation

The conditional expectation $\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$ is simply the updated mean vector $m_k$:

$$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}] = m_k$$

$$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}] = m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

where the Kalman Gain is $K_k = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$.


#### Finding the Conditional Variance

The conditional variance $\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$ is the updated error covariance matrix $P_k$:

$$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) = P_k$$

$$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) = (I - K_k H_k) P_k^-$$


The explicit closed-form expressions for the required quantities are:

$$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}] = m_k^- + P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1} (y^{\mathrm{obs}}_{k} - H_k m_k^-)$$

$$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) = P_k^- - P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1} H_k P_k^-$$

\begin{align*}
p(y^-_k\mid Y_k)=\mathscr N\bigl(h\,m_k,\; h^2 P_k + r\bigr).
\end{align*}

5. For suitable numerical values for $a,q,h,r$ animate the above to prior distribution and the posterior distribution.
