#Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$





###1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

 $\Omega = \{0,1,2,3\}$

 Let the $\mathfrak F$ be the power set of $\Omega$

 $\mathfrak F=\mathcal P(\Omega)
=\{\emptyset,\{0\},\{1\},\{2\},\{3\},\{0,1\},\ldots,\Omega\}.$

Assign each probabilities $P:\mathfrak F\to[0,1]$ as

$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$

###2. Define $X$ as a random variable on this space.

$
X:\Omega \to \mathbb{R}
$

$
X(\omega)=\omega, \qquad \omega \in \Omega
$

Since $\Omega=\{0,1,2,3\}$,

$
X(0)=0,\qquad
X(1)=1,\qquad
X(2)=2,\qquad
X(3)=3.
$

###3. Determine the probability mass function of $X$.


$$
p_X(x)=P(X=x),
\qquad x\in\{0,1,2,3\}.
$$

Hence,

$$
p_X(x)=
\begin{cases}
0.50, & x=0,\\[4pt]
0.30, & x=1,\\[4pt]
0.15, & x=2,\\[4pt]
0.05, & x=3,\\[4pt]
0, & \text{otherwise}.
\end{cases}
$$

###4. Compute the expectation $\mathbb E[X]$.
$
\mathbb{E}[X]
= \sum_{x} x\,P(X=x)
$

$
= 0(0.50) + 1(0.30) + 2(0.15) + 3(0.05)$

$
= 0.75
$

###5. Compute the variance $\mathrm{Var}(X)$.

$
\mathrm{Var}(X)=\mathbb{E}[X^2]-(\mathbb{E}[X])^2
$

$
\mathbb{E}[X^2]
= 0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05)
$

$
\mathbb{E}[X^2]
= 0 + 0.30 + 0.60 + 0.45 = 1.35
$

$
\mathrm{Var}(X)=1.35-(0.75)^2
$

$
\mathrm{Var}(X)=1.35-0.5625
$

$
\mathrm{Var}(X)=0.7875
$

###6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
$
P(X \ge 1) = P(X=1) + P(X=2) + P(X=3)
$

$
P(X \ge 1) = 0.30 + 0.15 + 0.05 = 0.50
$

$
P(X \ge 2) = P(X=2) + P(X=3)
$

$
P(X \ge 2) = 0.15 + 0.05 = 0.20
$

###7. Briefly explain the practical meaning of the expectation and variance in this setting.

The expectation

$$
\mathbb{E}[X]=0.75
$$

represents the average number of emergency transformer failures expected per day over a long period of time.

The variance

$$
\mathrm{Var}(X)=0.7875
$$

measures how much the daily number of transformer failures fluctuates around the mean value. A larger variance indicates greater variability and less consistency in the number of failures reported each day.
