# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.


Bertrand's paradox shows that different mathematically natural definitions of a ``random chord'' lead to different probabilities.

Let the circle have radius $R$.

The side length of the inscribed equilateral triangle is

$$
s=\sqrt{3}R
$$

A chord at perpendicular distance $d$ from the center has length

$$
L=2\sqrt{R^2-d^2}
$$

We require

$$
2\sqrt{R^2-d^2}>\sqrt{3}R
$$

Squaring both sides,

$$
4(R^2-d^2)>3R^2
$$

$$
R^2>4d^2
$$

$$
d<\frac{R}{2}
$$

Thus, the chord is longer than the side of the inscribed equilateral triangle iff its midpoint lies within distance \(R/2\) of the center.



##Model 1: Random Endpoints on the Circumference

Choose two points independently and uniformly on the circumference and join them.

Fix one endpoint and let $\theta\in[0,\pi]$ be the subtended central angle.

The chord length is

$$
L=2R\sin\left(\frac{\theta}{2}\right)
$$

We require

$$
2R\sin\left(\frac{\theta}{2}\right)>\sqrt{3}R
$$

$$
\sin\left(\frac{\theta}{2}\right)>\frac{\sqrt{3}}{2}
$$

Hence,

$$
\frac{\theta}{2}>\frac{\pi}{3}
$$

$$
\theta>\frac{2\pi}{3}
$$

Since $\theta$ is uniformly distributed on $[0,\pi]$,

$$
P(L>\sqrt{3}R)
=
\frac{\pi-\frac{2\pi}{3}}{\pi}
=
\frac{1}{3}
$$

Therefore,

$$
{P=\frac{1}{3}}
$$



##Model 2: Midpoint Uniform Along a Radius

Choose a random radius direction, then choose the midpoint of the chord uniformly along that radius.

In this model, the distance \(d\) from the center is uniformly distributed on \([0,R]\).

The chord is sufficiently long iff

$$
d<\frac{R}{2}
$$

Thus,

$$
P
=
\frac{R/2}{R}
=
\frac{1}{2}
$$

Therefore,

$$
{P=\frac{1}{2}}
$$



##Model 3: Midpoint Uniformly Distributed in the Disk

Choose the midpoint of the chord uniformly over the area of the disk.

The chord is sufficiently long iff the midpoint lies inside the concentric circle of radius \(R/2\).

Hence,

$$
P
=
\frac{\pi(R/2)^2}{\pi R^2}
=
\frac{1}{4}
$$

Therefore,

$$
{P=\frac{1}{4}}
$$

###Conclusion

Different mathematically natural models of a ``random chord'' lead to different probabilities:

$$
{\frac13,\quad \frac12,\quad \frac14}
$$

Thus, Bertrand's paradox illustrates that the phrase ``choose a chord at random'' is ambiguous unless the probability model is explicitly specified.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?


Let

$$
S=\{BB,BG,GB,GG\}
$$

be the sample space, where:

- \(BB\): both children are boys  
- \(BG\): older child is a boy, younger is a girl  
- \(GB\): older child is a girl, younger is a boy  
- \(GG\): both children are girls  

Since each child is independently equally likely to be a boy or a girl,

$$
P(BB)=P(BG)=P(GB)=P(GG)=\frac14
$$

We are given that at least one child is a girl.

Thus, the possible outcomes are

$$
A=\{BG,GB,GG\}
$$

Among these three equally likely outcomes, only one corresponds to both children being girls:

$$
\{GG\}
$$

Therefore,

$$
P(\text{both girls} \mid \text{at least one girl})
=
\frac{1}{3}
$$

Hence,

$$
{\frac13}
$$


# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

7. Compute the joint probabilities
   $$
   P(X=0,Y=1),\quad P(X=0,Y=2),\quad P(X=1,Y=3),\quad P(X=1,Y=4).
   $$

8. Compute the conditional probabilities
   $$
   P(X=1\mid Y=3),\qquad P(X=1\mid Y=4),\qquad P(Y=3\mid X=1),\qquad P(Y=4\mid X=1).
   $$

9. Interpret the difference between conditioning on $X$ and conditioning on $Y$.

10. Explain, in words, what information is available in $\sigma(X)$ but not in $\sigma(Y)$, and what information is available in $\sigma(Y)$ but not in $\sigma(X)$.

11. Give a Python based visualization of the model.

##1.

$$
\Omega=\{G_1,G_2,D_1,D_2\}.
$$

The $\sigma$-algebra is the power set

$$
\mathfrak F=\mathfrak P(\Omega).
$$

The probability measure is

$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20.
$$

Thus,

$$
(\Omega,\mathfrak F,P).
$$

##2.
We have

$$
X^{-1}(0)=\{G_1,G_2\},\quad X^{-1}(1)=\{D_1,D_2\}.
$$

Hence

$$
\sigma(X)=\{\emptyset,\Omega,\{G_1,G_2\},\{D_1,D_2\}\}.
$$

##3.
$\sigma$-algebra Generated by $Y$

Since $Y$ assigns distinct values to each outcome,

$$
Y^{-1}(1)=\{G_1\},\;
Y^{-1}(2)=\{G_2\},\;
Y^{-1}(3)=\{D_1\},\;
Y^{-1}(4)=\{D_2\}.
$$

Thus

$$
\sigma(Y)=\mathfrak P(\Omega).
$$

##4.
Relation Between $\sigma(X)$ and $\sigma(Y)$

Since

$$
\sigma(X)\subseteq \mathfrak P(\Omega)=\sigma(Y),
$$

we have

$$
\sigma(X)\subseteq \sigma(Y).
$$

So $Y$ contains more information because it identifies both machine and quality, while $X$ only identifies quality.

##5. Marginal Distribution of $X$

$$
P(X=0)=P(G_1)+P(G_2)=0.70
$$

$$
P(X=1)=P(D_1)+P(D_2)=0.30
$$

##6.
Marginal Distribution of $Y$

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20.
$$

##7.
 Joint Probabilities

$$
P(X=0,Y=1)=0.50
$$

$$
P(X=0,Y=2)=0.20
$$

$$
P(X=1,Y=3)=0.10
$$

$$
P(X=1,Y=4)=0.20
$$

##8.
Conditional Probabilities

$$
P(X=1\mid Y=3)=\frac{0.10}{0.10}=1
$$

$$
P(X=1\mid Y=4)=\frac{0.20}{0.20}=1
$$

$$
P(Y=3\mid X=1)=\frac{0.10}{0.30}=\frac{1}{3}
$$

$$
P(Y=4\mid X=1)=\frac{0.20}{0.30}=\frac{2}{3}
$$

##9.
Interpretation of Conditioning

Conditioning on $Y$ gives complete information about both machine and quality, so outcomes become deterministic.

Conditioning on $X$ only tells whether the product is good or defective, leaving uncertainty about the machine.

##10.
Information in $\sigma(X)$ and $\sigma(Y)$

$\sigma(X)$ only distinguishes:

- good vs defective

$\sigma(Y)$ distinguishes:

- good/defective AND machine identity

Thus,

$$
\sigma(X)\subset \sigma(Y),
$$

and $Y$ is strictly more informative.
